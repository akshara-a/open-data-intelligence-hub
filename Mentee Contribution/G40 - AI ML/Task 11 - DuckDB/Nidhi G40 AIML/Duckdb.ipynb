```python
import pandas as pd
import pyarrow
import duckdb

print(pd.__version__)
print(pyarrow.__version__)
```

```python
data = {
    "employee_id": [1, 2, 3, 4, 5, 6, 7, 8, 9],
    "name": [
        "Nidhi",
        "Atharv",
        "Deepak",
        "Priya",
        "Aishwarya",
        "Akanksha",
        "Mitali",
        "Mansi",
        "Varun"
    ],
    "department": [
        "HR",
        "IT",
        "HR",
        "Finance",
        "HR",
        "Finance",
        "Accountant",
        "Accountant",
        "IT"
    ],
    "salary": [
        30000,
        50000,
        80000,
        70000,
        60000,
        100000,
        78000,
        85000,
        75000
    ],
    "city": [
        "Pune",
        "Mumbai",
        "Delhi",
        "Pune",
        "Nashik",
        "Mumbai",
        "Delhi",
        "Pune",
        "Bengaluru"
    ]
}

df = pd.DataFrame(data)

df.to_parquet("employees.parquet", index=False)

print("Parquet file created successfully.")
```

### Read Parquet Using DuckDB

```python
result = duckdb.sql("""
SELECT *
FROM read_parquet('employees.parquet')
""").df()

print("\nAll Employee Records")
print(result)
```
