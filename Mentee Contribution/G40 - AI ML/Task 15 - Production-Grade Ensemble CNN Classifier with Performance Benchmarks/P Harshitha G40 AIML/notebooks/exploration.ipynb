```python
import sys
sys.path.append('../src')
import matplotlib.pyplot as plt
import numpy as np
from data_loader import get_data_loaders, CIFAR10_CLASSES
from preprocessing import get_base_transform, denormalize_image

# Load dataset loaders
train_loader, val_loader, test_loader, classes = get_data_loaders(data_dir='../data')
print(f"Classes: {classes}")
print(f"Train samples: {len(train_loader.dataset)}")
print(f"Val samples:   {len(val_loader.dataset)}")
print(f"Test samples:  {len(test_loader.dataset)}")
```

```python
# Visualize 10 sample images (one for each class)
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
found_classes = {}

for images, labels in train_loader:
    for img, lbl in zip(images, labels):
        lbl_idx = lbl.item()
        if lbl_idx not in found_classes:
            found_classes[lbl_idx] = img
        if len(found_classes) == 10:
            break
    if len(found_classes) == 10:
        break

for idx, ax in enumerate(axes.flat):
    img_t = denormalize_image(found_classes[idx])
    img_np = img_t.permute(1, 2, 0).numpy()
    ax.imshow(img_np)
    ax.set_title(CIFAR10_CLASSES[idx].capitalize())
    ax.axis('off')

plt.tight_layout()
plt.show()
```
