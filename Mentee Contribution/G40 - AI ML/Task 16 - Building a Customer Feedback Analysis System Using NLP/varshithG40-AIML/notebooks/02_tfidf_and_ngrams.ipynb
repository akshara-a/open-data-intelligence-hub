# 02 - Bag of Words, N-Grams & TF-IDF

This notebook covers how text is converted into numerical vectors:
1. Bag of Words (BoW) & its limitations
2. Term Frequency (TF) & Inverse Document Frequency (IDF)
3. Unigrams vs Bigrams (N-grams)
4. Building a TF-IDF Matrix with scikit-learn


```python
import sys
sys.path.append('..')

import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

## 1. Bag of Words Representation
Counting word frequencies in each document:


```python
corpus = [
    "payment failed again",
    "application is very slow",
    "support team was helpful",
    "payment was very smooth"
]

bow_vectorizer = CountVectorizer()
bow_matrix = bow_vectorizer.fit_transform(corpus)
pd.DataFrame(bow_matrix.toarray(), columns=bow_vectorizer.get_feature_names_out(), index=corpus)
```

## 2. Why TF-IDF?
Words appearing in every document (e.g. 'application', 'the') receive lower weights, while distinguishing terms receive higher weights:


```python
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)
pd.DataFrame(tfidf_matrix.toarray().round(3), columns=tfidf_vectorizer.get_feature_names_out(), index=corpus)
```

## 3. Adding N-Grams (Bigrams & Trigrams)
Single words (unigrams) lose contextual pairing like *'not good'* or *'payment failed'*. Bigrams preserve word pairs:


```python
ngram_vec = TfidfVectorizer(ngram_range=(1, 2))
ngram_matrix = ngram_vec.fit_transform(corpus)
print(f"Total vocabulary terms with Bigrams: {len(ngram_vec.get_feature_names_out())}")
pd.DataFrame(ngram_matrix.toarray().round(2), columns=ngram_vec.get_feature_names_out(), index=corpus)
```
