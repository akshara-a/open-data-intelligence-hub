# 03 - Sentiment Analysis with Logistic Regression

In this notebook, we train a supervised Sentiment Classifier on customer feedback:
1. Train / Test Split
2. Pipeline with TF-IDF + Logistic Regression
3. Precision, Recall, F1 Score & Confusion Matrix
4. Inspecting Model Feature Coefficients (Explainability)


```python
import sys
sys.path.append('..')

import pandas as pd
from sklearn.model_selection import train_test_split
from src.sentiment import SentimentClassifier
```

## 1. Load Dataset


```python
df = pd.read_csv("../data/feedback.csv")
print(f"Total samples: {len(df)}")
df[["feedback", "sentiment"]].head(8)
```

## 2. Train / Test Split


```python
train_texts, test_texts, train_labels, test_labels = train_test_split(
    df["feedback"].tolist(), df["sentiment"].tolist(), test_size=0.25, random_state=42, stratify=df["sentiment"].tolist()
)
print(f"Training samples: {len(train_texts)}, Test samples: {len(test_texts)}")
```

## 3. Train Sentiment Classifier


```python
model = SentimentClassifier(ngram_range=(1, 2))
model.fit(train_texts, train_labels)
print("Classes:", model.classes_)
```

## 4. Evaluate Performance


```python
eval_results = model.evaluate(test_texts, test_labels)
print(f"Test Accuracy: {eval_results['accuracy'] * 100:.2f}%")
print(f"Test F1 Score: {eval_results['f1_score'] * 100:.2f}%")

pd.DataFrame(eval_results['classification_report']).transpose()
```

## 5. Model Explainability: Top Informative Features
Which n-grams most strongly drive positive, negative, and neutral predictions?


```python
top_feats = model.get_top_features(n_top=6)
for sent, feats in top_feats.items():
    print(f"\n--- {sent.upper()} ---")
    for word, weight in feats:
        print(f"  {word:<18}: {weight:+.3f}")
```
