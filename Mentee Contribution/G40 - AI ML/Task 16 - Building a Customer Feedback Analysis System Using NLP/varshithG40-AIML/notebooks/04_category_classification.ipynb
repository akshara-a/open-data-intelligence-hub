# 04 - Multi-Label Customer Feedback Classification

Customer reviews often mention multiple concerns in a single sentence:
> *"The app is very slow and payment keeps failing."*

This contains **both** `Performance` and `Payment`.
Here we implement:
1. MultiLabelBinarizer
2. One-vs-Rest (OvR) Classification with Logistic Regression
3. Micro F1, Macro F1, and Hamming Loss Evaluation


```python
import sys
sys.path.append('..')

import pandas as pd
from src.category_classifier import MultiLabelCategoryClassifier
```

## 1. Load Multi-Label Dataset


```python
df = pd.read_csv("../data/feedback.csv")
df[["feedback", "categories"]].head(10)
```

## 2. Train Multi-Label Classifier


```python
cat_model = MultiLabelCategoryClassifier(threshold=0.35)
cat_model.fit(df["feedback"].tolist(), df["categories"].tolist())
print("Supported categories:", cat_model.classes_)
```

## 3. Test on Multi-Topic Feedback


```python
test_cases = [
    "The application is very slow and payment keeps failing.",
    "Customer support solved my billing issue very quickly.",
    "Can you please add dark mode support in the app?",
    "Fingerprint login fails after the latest update."
]

for text in test_cases:
    preds = cat_model.predict_one(text)
    scores = cat_model.predict_proba_one(text)
    print(f"\nText:       '{text}'")
    print(f"Predicted:  {preds}")
    top_scores = {k: v for k, v in scores.items() if v >= 0.25}
    print(f"Scores:     {top_scores}")
```

## 4. Multi-Label Evaluation Metrics
For multi-label tasks, we evaluate **Micro F1**, **Macro F1**, and **Hamming Loss**:


```python
metrics = cat_model.evaluate(df["feedback"].tolist(), df["categories"].tolist())
print(f"Micro F1:     {metrics['micro_f1']*100:.2f}%")
print(f"Macro F1:     {metrics['macro_f1']*100:.2f}%")
print(f"Hamming Loss: {metrics['hamming_loss']:.4f}")
```
