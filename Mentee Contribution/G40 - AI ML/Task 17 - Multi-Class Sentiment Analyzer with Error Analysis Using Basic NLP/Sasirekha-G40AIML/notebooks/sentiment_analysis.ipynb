# Multi-Class Sentiment Analyzer
End-to-end exploration, training, evaluation and error analysis using the supplied `Tweets.csv` dataset.


```python
import pandas as pd
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from src.model import build_pipeline

ROOT = Path('..')
df = pd.read_csv(ROOT / 'data/raw/Tweets.csv').dropna(subset=['text','sentiment'])
df.head()
```

```python
df.info()
df['sentiment'].value_counts()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df['text'].astype(str), df['sentiment'].astype(str),
    test_size=0.20, random_state=42, stratify=df['sentiment']
)
model = build_pipeline()
model.fit(X_train, y_train)
pred = model.predict(X_test)
print(classification_report(y_test, pred))
```
