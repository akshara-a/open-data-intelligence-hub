```python
#Create the Data Frame
import pandas as pd 
data={
    "employee_id":[1,2,3,4,5],
    "name": ["Asha", "Rahul", "Neha", "Vikram", "Priya"],
    "department": ["IT", "HR", "IT", "Finance", "HR"],
    "salary": [60000, 45000, 70000, 55000, 48000]
}
employees_df = pd.DataFrame(data)
print(employees_df)
```

```python
# save as parquet
employees_df.to_parquet(
    "employees.parquet",
    index=False
)

print("employees.parquet created successfully!")
#(index=False because it do not save the dataframe index)
```

```python
# Read the Parquet file
loaded_df=pd.read_parquet("employees.parquet")
print(loaded_df)
```

```python
# Employees earing more than ₹50,000
high_salary=loaded_df[loaded_df["salary"]>50000]
print("Employees with salary greater than 50000:")
print(high_salary)
```

```python
# Average salary
average_salary=loaded_df["salary"].mean()
print("Average Salary:", average_salary)
```

```python
# Number of employees in each department
department_count=loaded_df["department"].value_counts()
print("Employees in each department:")
print(department_count)
```

```python
# Save filtered employees
# save employees earning more than 50,000
high_salary.to_parquet(
    "high_salary_employees.parquet",
    index=False
)

print("high_salary_employees.parquet created successfully!")
```

```python
# Bonus
bonus_df = pd.read_parquet(
    "employees.parquet",
    columns=["name", "salary"]
)
print(bonus_df)
```
