```python
import pandas as pd
```

###   Create the DataFrame

```python
data = {
    "employee_id": [1,2,3,4,5,6,7,8,9],
    "name": ["Nidhi","Atharv","Deepak","Priya","Aishwarya","Akanksha","Mitali","Mansi","Varun"],
    "department": ["HR","IT","HR","FINANCE","HR","FINANCE","ACCOUNTANT","ACCOUNTANT","IT"],
    "salary": [30000,50000,80000,70000,60000,100000,78000,85000,75000]
}

employees_df = pd.DataFrame(data)

print("All Employee Records")
print(employees_df)
```

###  Save as Parquet

```python

employees_df.to_parquet("employees.parquet", index=False)

print("\nemployees.parquet file created successfully.")
```

### Read the Parquet File

```python
loaded_df = pd.read_parquet("employees.parquet")

print("\nData Read from employees.parquet")
print(loaded_df)
```

### Perform Basic Analysis

```python
# 1. Employees with salary greater than 50000
high_salary = loaded_df[loaded_df["salary"] > 50000]

print("\nEmployees with Salary Greater than 50000")
print(high_salary)
```

```python
# 2. Average Salary
average_salary = loaded_df["salary"].mean()

print("\nAverage Salary:")
print(average_salary)
```

```python
# 3. Employee Count in Each Department
department_count = loaded_df["department"].value_counts()

print("\nEmployee Count by Department:")
print(department_count)
```

### Save Filtered Data

```python
high_salary.to_parquet("high_salary_employees.parquet", index=False)

print("\nhigh_salary_employees.parquet created successfully.")
```
