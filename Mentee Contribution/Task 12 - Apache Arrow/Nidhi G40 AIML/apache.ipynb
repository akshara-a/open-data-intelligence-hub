# Apache Arrow Assignment Using Python

### Import Libraries

```python
import pyarrow as pa
import pyarrow.compute as pc
import pyarrow.parquet as pq
import pyarrow.ipc as ipc
import pandas as pd
```

### Create an Arrow Table

```python
data = {
    "employee_id": [1, 2, 3, 4, 5, 6, 7, 8, 9],
    "name": [
        "Nidhi",
        "Atharv",
        "Deepak",
        "Priya",
        "Aishwarya",
        "Akanksha",
        "Mitali",
        "Mansi",
        "Varun"
    ],
    "department": [
        "HR",
        "IT",
        "HR",
        "Finance",
        "HR",
        "Finance",
        "Accountant",
        "Accountant",
        "IT"
    ],
    "salary": [
        30000,
        50000,
        80000,
        70000,
        60000,
        100000,
        78000,
        85000,
        75000
    ],
    "city": [
        "Pune",
        "Mumbai",
        "Delhi",
        "Pune",
        "Nashik",
        "Mumbai",
        "Delhi",
        "Pune",
        "Bengaluru"
    ]
}

employee_table = pa.table(data)

print("Task 1: Arrow Table")
print(employee_table)
```

### Display the Schema

```python

print("\nTask 2: Schema")
print(employee_table.schema)

print("\nData Types:")
print("employee_id:", employee_table.schema.field("employee_id").type)
print("name:", employee_table.schema.field("name").type)
print("salary:", employee_table.schema.field("salary").type)
```

### Inspect the Table

```python
print("\nTask 3: Inspect Table")

print("Rows:", employee_table.num_rows)
print("Columns:", employee_table.num_columns)
print("Column names:", employee_table.column_names)

print("\nName Column:")
print(employee_table.column("name"))

print("\nFirst Three Rows:")
print(employee_table.slice(0, 3))
```

### Select Specific Columns

```python
selected_table = employee_table.select(
    ["name", "department", "salary"]
)

print("\nTask 4: Selected Columns")
print(selected_table)
```

### Filter Salary > 50000

```python
salary_filter = pc.greater(
    employee_table["salary"],
    50000
)

high_salary_table = employee_table.filter(
    salary_filter
)

print("\nTask 5: Salary Greater Than 50000")
print(high_salary_table)
```

### Filter IT Department

```python
department_filter = pc.equal(
    employee_table["department"],
    "IT"
)

it_employees = employee_table.filter(
    department_filter
)

print("\nTask 6: IT Department Employees")
print(it_employees)
```

###  Perform Calculations

```python
salary_column = employee_table["salary"]

print("\nTask 7: Salary Calculations")

print(
    "Average Salary:",
    pc.mean(salary_column).as_py()
)

print(
    "Maximum Salary:",
    pc.max(salary_column).as_py()
)

print(
    "Minimum Salary:",
    pc.min(salary_column).as_py()
)

print(
    "Total Salary:",
    pc.sum(salary_column).as_py()
)
```

### Add Bonus Column

```python
bonus_column = pc.multiply(
    employee_table["salary"],
    0.10
)

employee_table = employee_table.append_column(
    "bonus",
    bonus_column
)

print("\nTask 8: Table with Bonus")
print(employee_table)
```

### Convert Arrow to Pandas

```python
employee_df = employee_table.to_pandas()

print("\nTask 9: Arrow to Pandas")
print(employee_df)
```

### Convert Pandas to Arrow

```python
new_arrow_table = pa.Table.from_pandas(
    employee_df,
    preserve_index=False
)

print("\nTask 10: Pandas to Arrow")
print(new_arrow_table)
```

### Save as Parquet

```python
pq.write_table(
    employee_table,
    "employees.parquet"
)

print("\nTask 11: Parquet file created successfully.")
```

###  Read Parquet File

```python
loaded_table = pq.read_table(
    "employees.parquet"
)

print("\nTask 12: Loaded Parquet File")
print(loaded_table)
```

###  Save as Arrow IPC File

```python
with ipc.new_file(
    "employees.arrow",
    employee_table.schema
) as writer:

    writer.write_table(employee_table)

print("\nTask 13: Arrow IPC file created successfully.")
```

###  Read Arrow IPC File

```python
with ipc.open_file("employees.arrow") as reader:

    ipc_table = reader.read_all()

print("\nTask 14: Loaded Arrow IPC File")
print(ipc_table)
```
