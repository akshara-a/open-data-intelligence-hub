```python
from pathlib import Path

DATA_DIR = Path("../data/casting_512x512")

DEFECTIVE_DIR = DATA_DIR / "def_front"
OK_DIR = DATA_DIR / "ok_front"

print("Defective folder:", DEFECTIVE_DIR)
print("OK folder:", OK_DIR)

print("Defective exists:", DEFECTIVE_DIR.exists())
print("OK exists:", OK_DIR.exists())
```

```python
defective_images = list(DEFECTIVE_DIR.glob("*"))
ok_images = list(OK_DIR.glob("*"))

print("Defective images:", len(defective_images))
print("Non-defective images:", len(ok_images))
print("Total images:", len(defective_images) + len(ok_images))
```

```python
from PIL import Image

sample_image = Image.open(defective_images[0])

print("Image size:", sample_image.size)
print("Image mode:", sample_image.mode)
```

```python
import matplotlib.pyplot as plt
from PIL import Image

fig, axes = plt.subplots(2, 4, figsize=(14, 7))

# Defective samples
for i in range(4):
    image = Image.open(defective_images[i])
    axes[0, i].imshow(image)
    axes[0, i].set_title("Defective")
    axes[0, i].axis("off")

# Non-defective samples
for i in range(4):
    image = Image.open(ok_images[i])
    axes[1, i].imshow(image)
    axes[1, i].set_title("Non-defective")
    axes[1, i].axis("off")

plt.tight_layout()
plt.show()
```

```python
total = len(defective_images) + len(ok_images)

defective_percentage = len(defective_images) / total * 100
ok_percentage = len(ok_images) / total * 100

print(f"Defective:     {len(defective_images):,} ({defective_percentage:.2f}%)")
print(f"Non-defective: {len(ok_images):,} ({ok_percentage:.2f}%)")
print(f"Total:         {total:,}")
```

```python
labels = ["Defective", "Non-defective"]
counts = [len(defective_images), len(ok_images)]

plt.figure(figsize=(7, 5))
plt.bar(labels, counts)
plt.ylabel("Number of Images")
plt.title("Casting Dataset Class Distribution")
plt.show()
```

```python
print("========== DATASET SUMMARY ==========")
print(f"Dataset location : {DATA_DIR}")
print(f"Image size       : {sample_image.size}")
print(f"Defective images : {len(defective_images):,}")
print(f"OK images        : {len(ok_images):,}")
print(f"Total images     : {total:,}")
print(f"Defective %      : {defective_percentage:.2f}%")
print(f"Non-defective %  : {ok_percentage:.2f}%")
```
