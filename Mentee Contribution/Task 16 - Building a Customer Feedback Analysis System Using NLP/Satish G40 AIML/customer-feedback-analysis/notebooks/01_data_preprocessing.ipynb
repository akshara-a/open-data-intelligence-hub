# 01. Data Preprocessing and EDA

This notebook loads the customer feedback dataset, checks for missing values, and explores sentiment and category distributions.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

project_root = Path.cwd().resolve().parent
data_path = project_root / 'data' / 'feedback.csv'
df = pd.read_csv(data_path)
df.head()
```

```python
print(df.info())
print('\nMissing values:')
print(df.isnull().sum())
```

```python
df['feedback_length'] = df['feedback'].str.len()
sns.histplot(df['feedback_length'], bins=20, kde=True)
plt.title('Feedback Length Distribution')
plt.xlabel('Length of feedback')
plt.ylabel('Count')
plt.show()
```

```python
sentiment_counts = df['sentiment'].value_counts()
sns.barplot(x=sentiment_counts.index, y=sentiment_counts.values, palette='Set2')
plt.title('Sentiment Distribution')
plt.xlabel('Sentiment')
plt.ylabel('Count')
plt.show()
```

```python
# Split multiple categories before plotting
all_categories = []
for categories in df['categories'].dropna():
    all_categories.extend(str(categories).split(','))
category_counts = pd.Series(all_categories).str.strip().str.lower().value_counts()

sns.barplot(x=category_counts.values, y=category_counts.index, palette='viridis')
plt.title('Category Distribution')
plt.xlabel('Count')
plt.ylabel('Category')
plt.show()
```
