```python
def clean_text(text):


    text = text.lower()

    text = re.sub(r"[^a-zA-Z\s]", "", text)

    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
sample_text = "THE    Payment is Failing!!!"

cleaned_text = clean_text(sample_text)

print("Original:")
print(sample_text)

print("\nCleaned:")
print(cleaned_text)
```

```python
df["cleaned_feedback"] = df["feedback"].apply(clean_text)

display(
    df[
        ["feedback", "cleaned_feedback"]
    ].head(10)
)
```

```python
def tokenize_text(text):

    return word_tokenize(text)
```

```python
sample_text = "the payment is failing"

tokens = tokenize_text(sample_text)

print("Original text:")
print(sample_text)

print("\nTokens:")
print(tokens)
```

```python
df["tokens"] = df["cleaned_feedback"].apply(tokenize_text)

display(
    df[
        ["cleaned_feedback", "tokens"]
    ].head(10)
)
```

```python
stop_words = set(stopwords.words("english"))

print("Number of stopwords:", len(stop_words))

print("\nSample stopwords:")
print(list(stop_words)[:20])
```

```python
def remove_stopwords(tokens):
    """
    Remove common English stopwords.
    """

    return [
        word
        for word in tokens
        if word not in stop_words
    ]
```

```python
df["tokens_no_stopwords"] = df["tokens"].apply(
    remove_stopwords
)

display(
    df[
        [
            "tokens",
            "tokens_no_stopwords"
        ]
    ].head(10)
)
```

```python
stemmer = PorterStemmer()
def stem_words(tokens):

    return [
        stemmer.stem(word)
        for word in tokens
    ]
```

```python
df["stemmed_tokens"] = df["tokens_no_stopwords"].apply(
    stem_words
)

display(
    df[
        [
            "tokens_no_stopwords",
            "stemmed_tokens"
        ]
    ].head(10)
)
```

```python
lemmatizer = WordNetLemmatizer()
def lemmatize_words(tokens):

    return [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]
```

```python
df["lemmatized_tokens"] = df["tokens_no_stopwords"].apply(
    lemmatize_words
)

display(
    df[
        [
            "tokens_no_stopwords",
            "lemmatized_tokens"
        ]
    ].head(10)
)
```

```python
df["processed_text"] = df["lemmatized_tokens"].apply(
    lambda tokens: " ".join(tokens)
)

display(
    df[
        [
            "feedback",
            "processed_text",
            "sentiment",
            "category"
        ]
    ].head(10)
)
```
