## 2. Tokenization
Breaking text into individual semantic tokens (words) so algorithms can learn relationships:


```python
text = "Payment keeps failing on checkout"
tokens = tokenize(text)
print("Tokens:", tokens)
```

## 3. The Danger of Over-Cleaning (Negation Preservation)
Why we should **not** blindly remove all stop words:
Consider: *'This application is not good'*
If 'not' is removed, the meaning completely flips to positive!


```python
sample = ["this", "application", "is", "not", "good"]
blind_removal = remove_stopwords(sample, preserve_negations=False)
preserving_negation = remove_stopwords(sample, preserve_negations=True)

print("Blind removal (Dangerous):   ", blind_removal)
print("Negation-preserving (Correct):", preserving_negation)
```

## 4. Stemming vs Lemmatization
- **Stemming** chops word endings using heuristic rules (can produce non-words like `studi`).
- **Lemmatization** uses vocabulary & morphological analysis to return valid dictionary root forms (`better` -> `good`).


```python
import pandas as pd
comparison = compare_stem_vs_lemma([
    "complaining", "payments", "better", "crashes", "failing", "studies", "connected"
])
pd.DataFrame(comparison)
```
