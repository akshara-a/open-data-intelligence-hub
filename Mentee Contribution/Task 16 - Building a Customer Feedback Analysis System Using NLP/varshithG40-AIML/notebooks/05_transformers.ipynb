## 1. Sentence Similarity & Finding Similar Complaints


```python
df = pd.read_csv("../data/feedback.csv")
engine = FeedbackSimilarityEngine()
engine.index_corpus(df)

query = "Unable to complete my credit card payment during checkout"
results = engine.find_similar(query, top_k=3)

print(f"Query: '{query}'\n")
for idx, match in enumerate(results, 1):
    print(f"{idx}. [{match['similarity_percentage']}] \"{match['feedback']}\"")
    print(f"   -> Sentiment: {match['sentiment']} | Categories: {match['categories']}")
```
