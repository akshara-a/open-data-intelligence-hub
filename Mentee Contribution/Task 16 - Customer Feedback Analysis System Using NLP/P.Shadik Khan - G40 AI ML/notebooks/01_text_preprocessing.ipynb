# 1. TEXT PREPROCESSING

This notebook demonstrates the preprocessing of customer feedback using Natural Language Processing (NLP).

## Objectives

- Load the customer feedback dataset
- Inspect the feedback data
- Convert text to lowercase
- Remove URLs
- Remove numbers
- Remove punctuation
- Remove extra spaces
- Prepare clean text for further NLP analysis

```python
import pandas as pd
import re
import string

print("Libraries imported successfully.")
```

## 1.1 Load Dataset

The customer feedback dataset is stored in `data/feedback.csv`.

We load the dataset using Pandas and inspect its contents.

```python
df = pd.read_csv("../data/feedback.csv")

print("Dataset loaded successfully.")
print("\nFirst 5 records:")
display(df.head())
```

## 1.2 Dataset Information

Let's examine the structure of the dataset, including the number of records, columns, and data types.

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

## 1.3 Inspect Customer Feedback

The feedback column contains the natural language text provided by customers.

```python
print("Customer Feedback:")
for i, feedback in enumerate(df["feedback"].head(10), start=1):
    print(f"{i}. {feedback}")
```

## 1.4 Convert Text to Lowercase

Converting text to lowercase ensures that words such as "Payment" and "payment" are treated consistently.

```python
df["lowercase_feedback"] = df["feedback"].str.lower()

display(
    df[["feedback", "lowercase_feedback"]].head(10)
)
```

## 1.5 Remove URLs

URLs are not useful for the basic customer feedback classification task, so they are removed during preprocessing.

```python
def remove_urls(text):
    return re.sub(r"http\S+|www\S+", "", text)

df["without_urls"] = df["lowercase_feedback"].apply(remove_urls)

display(
    df[["lowercase_feedback", "without_urls"]].head(10)
)
```

## 1.6 Remove Numbers

Numbers are removed because they are not required for the basic feedback classification process.

```python
def remove_numbers(text):
    return re.sub(r"\d+", "", text)

df["without_numbers"] = df["without_urls"].apply(remove_numbers)

display(
    df[["without_urls", "without_numbers"]].head(10)
)
```

## 1.7 Remove Punctuation

Punctuation marks are removed to simplify the text before applying machine learning techniques.

```python
def remove_punctuation(text):
    return text.translate(
        str.maketrans("", "", string.punctuation)
    )

df["without_punctuation"] = (
    df["without_numbers"].apply(remove_punctuation)
)

display(
    df[["without_numbers", "without_punctuation"]].head(10)
)
```

## 1.8 Remove Extra Spaces

Multiple spaces are replaced with a single space and leading or trailing spaces are removed.

```python
def remove_extra_spaces(text):
    return re.sub(r"\s+", " ", text).strip()

df["clean_feedback"] = (
    df["without_punctuation"].apply(remove_extra_spaces)
)

display(
    df[["feedback", "clean_feedback"]].head(10)
)
```

## 1.9 Final Preprocessed Dataset

The final cleaned feedback is now ready for further NLP processing such as tokenization, TF-IDF feature extraction, sentiment analysis, and category classification.

```python
final_columns = [
    "id",
    "feedback",
    "sentiment",
    "category",
    "clean_feedback"
]

display(df[final_columns].head(10))
```

## 1.10 Conclusion

Text preprocessing successfully cleaned the customer feedback.

The preprocessing steps included:

1. Lowercasing
2. URL removal
3. Number removal
4. Punctuation removal
5. Extra-space removal

The cleaned text can now be used for TF-IDF feature extraction and machine learning models.
