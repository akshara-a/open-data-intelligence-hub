```python
import pandas as pd
import matplotlib.pyplot as plt

from transformers import (
    pipeline,
    AutoTokenizer,
    AutoModelForSequenceClassification
)
```

## 5.1 Load Customer Feedback Dataset

Load the customer feedback dataset used in the previous NLP notebooks.

```python
df = pd.read_csv("../data/feedback.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

## 5.2 Inspect Customer Feedback

Display some customer feedback messages that will be used for
Transformer-based analysis.

```python
print("Sample Customer Feedback:")

for i, feedback in enumerate(df["feedback"].head(10), start=1):
    print(f"{i}. {feedback}")
```

## 5.3 Understanding Transformers

Transformers process text using tokenization and attention mechanisms.

The basic workflow is:

Customer Feedback
        |
        v
Transformer Tokenizer
        |
        v
Token IDs
        |
        v
Transformer Model
        |
        v
Classification
        |
        v
Sentiment

```python
model_name = "distilbert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(
    model_name)
print("Transformer tokenizer loaded successfully!")
print("Model:", model_name)
```

## 5.4 Transformer Tokenization

Tokenization converts customer feedback into tokens that can be
processed by the Transformer model.

```python
sample_text = "The application is very slow and payment keeps failing."

tokens = tokenizer.tokenize(sample_text)

print("Original Text:")
print(sample_text)

print("\nTokens:")
print(tokens)

print("\nNumber of Tokens:", len(tokens))
```

## 5.5 Convert Tokens to Token IDs

Transformers use numerical token IDs internally.

The tokenizer converts the text into numerical values that can be
processed by the model.

```python
encoded_text = tokenizer(
    sample_text,
    return_tensors="pt"
)

print("Token IDs:")
print(encoded_text["input_ids"])

print("\nAttention Mask:")
print(encoded_text["attention_mask"])
```

## 5.6 Load Pre-trained Sentiment Transformer

Load a pre-trained DistilBERT sentiment classification pipeline.

The model automatically predicts sentiment from customer feedback.

```python
sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

print("Sentiment transformer loaded successfully!")
```

## 5.7 Test Transformer Sentiment Analysis

Test the Transformer model with a few customer feedback examples.

```python
test_feedback = [
    "I love the new dashboard",
    "The application is very slow",
    "Payment keeps failing",
    "Support was very helpful"
]

transformer_results = sentiment_pipeline(
    test_feedback
)

results_df = pd.DataFrame({
    "Feedback": test_feedback,
    "Transformer Result": [
        result["label"]
        for result in transformer_results
    ],
    "Confidence": [
        result["score"]
        for result in transformer_results
    ]
})

results_df
```

## 5.8 Analyze Entire Customer Feedback Dataset

Apply the Transformer sentiment model to the complete feedback dataset.

```python
all_feedback = df["feedback"].tolist()

all_transformer_results = sentiment_pipeline(
    all_feedback
)

df["transformer_sentiment"] = [
    result["label"]
    for result in all_transformer_results
]

df["transformer_confidence"] = [
    result["score"]
    for result in all_transformer_results
]

df[
    [
        "feedback",
        "sentiment",
        "transformer_sentiment",
        "transformer_confidence"
    ]
].head(10)
```

## 5.9 Transformer Sentiment Distribution

Analyze the sentiment predictions generated by the Transformer model.

```python
transformer_distribution = (
    df["transformer_sentiment"]
    .value_counts()
)

print("Transformer Sentiment Distribution:")
print(transformer_distribution)

transformer_distribution.plot(
    kind="bar",
    title="Transformer Sentiment Distribution"
)

plt.xlabel("Transformer Sentiment")
plt.ylabel("Number of Feedback Records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 5.10 Compare Dataset Sentiment with Transformer Predictions

Compare the original dataset sentiment labels with the sentiment
predicted by the Transformer model.

Note: The pre-trained DistilBERT model used here was trained on a
general sentiment dataset and was not specifically fine-tuned on
our customer feedback dataset.

```python
comparison = df[[
    "feedback",
    "sentiment",
    "transformer_sentiment",
    "transformer_confidence"
]].copy()

comparison.head(15)
```

## 5.11 Test New Customer Feedback

Use the Transformer model to analyze new customer feedback that was
not included in the dataset.

```python
new_feedback = [
    "The latest update is excellent",
    "The application is extremely slow",
    "Payment failed several times",
    "The support team solved my problem quickly",
    "I am unhappy with the new interface"
]

new_transformer_results = sentiment_pipeline(
    new_feedback
)

new_results = pd.DataFrame({
    "Customer Feedback": new_feedback,
    "Predicted Sentiment": [
        result["label"]
        for result in new_transformer_results
    ],
    "Confidence": [
        round(result["score"], 4)
        for result in new_transformer_results
    ]
})

display(new_results)
```

## 5.12 TF-IDF vs Transformer

The project uses two different NLP approaches.

### TF-IDF + Logistic Regression

- Easy to understand
- Fast
- Requires less computing power
- Provides a strong baseline
- Has limited contextual understanding

### Transformer

- More complex
- Requires more computing resources
- Understands context better
- Uses tokenization and attention
- Can capture deeper language relationships

The Transformer approach is demonstrated as an advanced extension
of the classical TF-IDF NLP pipeline.

```python
comparison = pd.DataFrame({
    "Approach": [
        "TF-IDF + Logistic Regression",
        "DistilBERT Transformer"
    ],
    "Type": [
        "Classical NLP",
        "Modern NLP"
    ],
    "Context Understanding": [
        "Limited",
        "Strong"
    ],
    "Computational Requirement": [
        "Low",
        "Higher"
    ],
    "Project Role": [
        "Baseline",
        "Advanced Extension"
    ]
})

comparison
```
