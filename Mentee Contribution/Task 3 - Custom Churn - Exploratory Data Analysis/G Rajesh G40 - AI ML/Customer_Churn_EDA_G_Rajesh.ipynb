```python
# Convert TotalCharges to numeric

df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

print("Conversion completed.")
```

```python
# Missing values after conversion

print(df.isnull().sum())
```

```python
# Remove missing values

df = df.dropna()

print("New Dataset Shape:", df.shape)
```

```python
# Save cleaned dataset

df.to_csv("dataset/Cleaned_Customer_Churn_Dataset_G_Rajesh.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
# Chart 1: Churn Count Plot

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))
sns.countplot(x="Churn", data=df)

plt.title("Customer Churn Count")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")

plt.savefig("charts/chart1_churn_count.png")
plt.show()
```

### Observation

- The chart shows the number of customers who churned and those who did not.
- It helps understand the overall class distribution of the target variable.

```python
# Chart 2: Churn Percentage Pie Chart

plt.figure(figsize=(6,6))

df["Churn"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Customer Churn Percentage")
plt.ylabel("")

plt.savefig("charts/chart2_churn_percentage.png")
plt.show()
```

### Observation

- The pie chart shows the percentage of customers who stayed and those who churned.
- This helps understand the overall churn ratio in the dataset.

```python
# Chart 3: Histogram of Monthly Charges

plt.figure(figsize=(8,5))

sns.histplot(df["MonthlyCharges"], bins=30, kde=True)

plt.title("Distribution of Monthly Charges")
plt.xlabel("Monthly Charges")
plt.ylabel("Number of Customers")

plt.savefig("charts/chart3_monthlycharges_histogram.png")
plt.show()
```

### Observation

- The histogram shows how monthly charges are distributed among customers.
- It helps identify whether most customers pay low, medium, or high monthly charges.

```python
# Chart 4: Box Plot of Monthly Charges

plt.figure(figsize=(6,4))

sns.boxplot(y=df["MonthlyCharges"])

plt.title("Box Plot of Monthly Charges")

plt.savefig("charts/chart4_monthlycharges_boxplot.png")
plt.show()
```

### Observation

- The box plot helps detect outliers in the Monthly Charges column.
- Any points outside the whiskers may represent unusually high or low monthly charges.

```python
# Chart 5: Contract Type Distribution

plt.figure(figsize=(7,5))

sns.countplot(x="Contract", data=df)

plt.title("Customer Contract Types")
plt.xlabel("Contract Type")
plt.ylabel("Count")

plt.savefig("charts/chart5_contract_distribution.png")
plt.show()
```

### Observation

- Month-to-month contracts are the most common.
- One-year and two-year contracts have fewer customers.

```python
# Chart 6: Contract Type vs Churn

plt.figure(figsize=(7,5))

sns.countplot(x="Contract", hue="Churn", data=df)

plt.title("Contract Type vs Churn")

plt.savefig("charts/chart6_contract_vs_churn.png")
plt.show()
```

### Observation

- Customers with month-to-month contracts show higher churn.
- Long-term contracts have lower churn rates.

```python
# Chart 7: Correlation Heatmap

# Select only numeric columns
numeric_df = df.select_dtypes(include=["number"])

plt.figure(figsize=(10, 8))

sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap")

plt.savefig("charts/chart7_correlation_heatmap.png")
plt.show()
```

### Observation

- The heatmap shows the correlation between numerical features.
- Strong positive or negative correlations help identify relationships among variables.

```python
# Chart 8: Scatter Plot

plt.figure(figsize=(8,5))

sns.scatterplot(x="tenure", y="MonthlyCharges", hue="Churn", data=df)

plt.title("Tenure vs Monthly Charges")

plt.savefig("charts/chart8_scatterplot.png")
plt.show()
```

### Observation

- The scatter plot shows the relationship between customer tenure and monthly charges.
- It also highlights how churn is distributed across different customers.

# Key Findings

## 1. What percentage of customers churned?

Approximately 26–27% of customers in the dataset churned, while the remaining customers stayed with the company.

## 2. Which customer group has the highest churn?

Customers with month-to-month contracts showed the highest churn rate.

## 3. Which numerical features seem related to churn?

Tenure, Monthly Charges, and Total Charges showed a noticeable relationship with customer churn.

## 4. Which categorical features seem related to churn?

Contract type, Internet Service, and Payment Method were related to churn behaviour.

## 5. Are there missing values or data quality issues?

Yes. The TotalCharges column contained missing/non-numeric values, which were cleaned during preprocessing.

## 6. Are there outliers in the dataset?

Yes. The box plot indicated that a few customers had unusually high monthly charges.

## 7. What are the top 3 factors affecting churn?

- Month-to-month contracts
- Higher monthly charges
- Lower customer tenure

## 8. What recommendations can reduce churn?

- Encourage customers to choose long-term contracts.
- Provide loyalty rewards for long-term customers.
- Offer discounts or personalized plans to customers with high monthly charges.

# Conclusion

This exploratory data analysis helped identify important patterns in customer churn. The analysis showed that contract type, tenure, and monthly charges have a significant influence on whether a customer is likely to churn. The visualizations and statistical analysis provide useful insights that can help businesses improve customer retention and reduce churn.

# Business Recommendations

- Encourage customers to switch from month-to-month to long-term contracts.
- Offer discounts and loyalty benefits to long-term customers.
- Monitor customers with high monthly charges and provide customized plans.
- Improve customer support for new customers during the first few months.
- Use churn prediction models in the future to identify customers at risk of leaving.
