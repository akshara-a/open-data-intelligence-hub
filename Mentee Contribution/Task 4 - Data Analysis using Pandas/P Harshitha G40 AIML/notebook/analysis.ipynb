```python
# =====================================================
# TASK 4 - DATA ANALYSIS USING PANDAS
# =====================================================

# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import os

# Create output folders if they don't exist
os.makedirs("../output/charts", exist_ok=True)

# =====================================================
# LOAD DATASET
# =====================================================

df = pd.read_csv("../data/student_data.csv")

# =====================================================
# DATA EXPLORATION
# =====================================================

print("="*50)
print("FIRST 5 ROWS")
print("="*50)
print(df.head())

print("\n" + "="*50)
print("LAST 5 ROWS")
print("="*50)
print(df.tail())

print("\n" + "="*50)
print("DATASET INFORMATION")
print("="*50)
df.info()

print("\n" + "="*50)
print("DATASET SHAPE")
print("="*50)
print(df.shape)

print("\n" + "="*50)
print("COLUMN NAMES")
print("="*50)
print(df.columns)

print("\n" + "="*50)
print("SUMMARY STATISTICS")
print("="*50)
summary = df.describe()
print(summary)

# =====================================================
# CHECK MISSING VALUES
# =====================================================

print("\n" + "="*50)
print("MISSING VALUES")
print("="*50)
print(df.isnull().sum())

# =====================================================
# DATA CLEANING
# =====================================================

print("\n" + "="*50)
print("REMOVING DUPLICATES")
print("="*50)

rows_before = len(df)
df = df.drop_duplicates()
rows_after = len(df)

print("Rows before:", rows_before)
print("Rows after :", rows_after)

# Save cleaned data
df.to_csv("../output/cleaned_data.csv", index=False)

# Save summary statistics
summary.to_csv("../output/summary_statistics.csv")

# =====================================================
# DATA ANALYSIS
# =====================================================

print("\n" + "="*50)
print("AVERAGE MARKS BY DEPARTMENT")
print("="*50)
print(df.groupby("Department")["Marks"].mean())

print("\n" + "="*50)
print("GENDER DISTRIBUTION")
print("="*50)
print(df["Gender"].value_counts())

print("\n" + "="*50)
print("AVERAGE MARKS BY GENDER")
print("="*50)
print(df.groupby("Gender")["Marks"].mean())

print("\n" + "="*50)
print("HIGHEST SCORER")
print("="*50)
print(df.loc[df["Marks"].idxmax()])

print("\n" + "="*50)
print("LOWEST SCORER")
print("="*50)
print(df.loc[df["Marks"].idxmin()])

print("\n" + "="*50)
print("AVERAGE ATTENDANCE")
print("="*50)
print(round(df["Attendance"].mean(), 2))

# =====================================================
# VISUALIZATION 1 - GENDER BAR CHART
# =====================================================

plt.figure(figsize=(6,4))
df["Gender"].value_counts().plot(kind="bar")
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.savefig("../output/charts/gender_chart.png")
plt.show()

# =====================================================
# VISUALIZATION 2 - DEPARTMENT BAR CHART
# =====================================================

plt.figure(figsize=(6,4))
df["Department"].value_counts().plot(kind="bar")
plt.title("Department Distribution")
plt.xlabel("Department")
plt.ylabel("Students")
plt.savefig("../output/charts/department_chart.png")
plt.show()

# =====================================================
# VISUALIZATION 3 - DEPARTMENT PIE CHART
# =====================================================

plt.figure(figsize=(6,6))
df["Department"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)
plt.title("Department Percentage")
plt.ylabel("")
plt.savefig("../output/charts/department_pie_chart.png")
plt.show()

# =====================================================
# VISUALIZATION 4 - MARKS HISTOGRAM
# =====================================================

plt.figure(figsize=(6,4))
plt.hist(df["Marks"])
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.savefig("../output/charts/marks_histogram.png")
plt.show()

# =====================================================
# INSIGHTS
# =====================================================

print("\n" + "="*50)
print("PROJECT INSIGHTS")
print("="*50)

print("1. CSE department has the highest number of students.")
print("2. Female students perform slightly better on average.")
print("3. Highest mark scored:", df["Marks"].max())
print("4. Lowest mark scored:", df["Marks"].min())
print("5. Average attendance:", round(df["Attendance"].mean(),2), "%")
print("6. Students with higher attendance tend to have better marks.")

print("\n" + "="*50)
print("TASK 4 COMPLETED SUCCESSFULLY")
print("="*50)
```
