```python
products = df[
    [
        "id",
        "name",
        "brand",
        "categories",
        "reviews.text",
        "reviews.rating",
        "reviews.username"
    ]
]

products.head()
```

```python
products["reviews.rating"] = products["reviews.rating"].fillna(0)

products["reviews.username"] = products["reviews.username"].fillna("Anonymous")
```

```python
products["description"] = (
    products["name"].fillna("") + " " +
    products["brand"].fillna("") + " " +
    products["categories"].fillna("") + " " +
    products["reviews.text"].fillna("")
)

products.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    stop_words="english"
)

tfidf_matrix = tfidf.fit_transform(products["description"])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(tfidf_matrix)
```

```python
similarity.shape
```

```python
def recommend(product_name):
    
    index = products[products["name"] == product_name].index[0]

    similarity_scores = list(enumerate(similarity[index]))

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    recommended_products = similarity_scores[1:6]

    for i, score in recommended_products:
        print(
            products.iloc[i]["name"],
            " | Similarity:",
            round(score, 3)
        )
```

```python
recommend("Kindle Paperwhite")
```

```python
unique_products = products.drop_duplicates(subset="name").reset_index(drop=True)

unique_products.head()
```

```python
unique_products["description"] = (
    unique_products["name"].fillna("") + " " +
    unique_products["brand"].fillna("") + " " +
    unique_products["categories"].fillna("")
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(stop_words="english")

tfidf_matrix = tfidf.fit_transform(unique_products["description"])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(tfidf_matrix)
```

```python
def recommend(product_name, top_n=5):

    matches = unique_products[
        unique_products["name"].str.contains(product_name, case=False, na=False)
    ]

    if matches.empty:
        print("Product not found.")
        return

    idx = matches.index[0]

    scores = list(enumerate(similarity[idx]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)

    print(f"\nRecommended products for '{matches.iloc[0]['name']}':\n")

    count = 0
    for i, score in scores[1:]:
        print(f"{count+1}. {unique_products.iloc[i]['name']} ({score:.2f})")
        count += 1
        if count == top_n:
            break
```

```python
recommend("Kindle")
```

```python
import pickle

pickle.dump(unique_products, open("products.pkl", "wb"))
pickle.dump(similarity, open("similarity.pkl", "wb"))
pickle.dump(tfidf, open("tfidf.pkl", "wb"))

print("Files saved successfully!")
```
